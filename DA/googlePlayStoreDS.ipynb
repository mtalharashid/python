```python
df_copy = df.copy()
```

```python
df_copy=df_copy.drop(df_copy.index[10472])
```

```python
df_copy[~df_copy['Reviews'].str.isnumeric()]
```

```python
### convert review to int
df_copy['Reviews']=df_copy['Reviews'].astype(int)
```

```python
df_copy.info()
```

```python
df_copy['Size'].unique()
```

```python
df_copy['Size'] = df_copy['Size'].str.replace('M', '000')
df_copy['Size']
```

```python
df_copy['Size'] = df_copy['Size'].str.replace('k', '')
df_copy['Size']
```

```python
df_copy['Size'] = df_copy['Size'].replace('Varies with device', np.nan)
df_copy['Size']=df_copy['Size'].astype(float)
```

```python
df_copy.info()
```

```python
df_copy['Size']
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
chars_to_remove = ['+',',','$']

cols_to_clean=['Installs', 'Price']

for item in chars_to_remove:
    for cols in cols_to_clean:
        df_copy[cols]=df_copy[cols].str.replace(item, '')
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs']=df_copy['Installs'].astype('int')
df_copy['Price']=df_copy['Price'].astype('float')
```

```python
df_copy.info()
```

```python
## now we are going to handel last update feature
df_copy['Last Updated'].unique()
```

```python
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
df_copy['Day'] = df_copy['Last Updated'].dt.day
df_copy['Month'] = df_copy['Last Updated'].dt.month
df_copy['Year'] = df_copy['Last Updated'].dt.year
```

```python
df_copy.info()
```

```python
df_copy.head()
```

```python
df_copy.to_csv('Google_cleaned.csv', index=False)
```

### Explore Data 


```python
numeric_features = [feature for feature in df_copy.columns if df_copy[feature].dtype !='0'] 
categorical_features = [feature for feature in df_copy.columns if df_copy[feature].dtype =='0'] 

print('We have {} numerical feature : {}'.format(len(numeric_features), numeric_features))
print('\nWe have {} categorical feature : {}'.format(len(categorical_features), categorical_features))
```

```python
## Proportion od count date in categorical columns
for col in categorical_features:
    print(df_copy[col].value_counts(normalize=True)*100)
    print('-------------------------------')
```

```python
print(df_copy[numeric_features].dtypes)
```

```python
numeric_features = [col for col in df_copy.columns if np.issubdtype(df_copy[col].dtype, np.number)]


plt.figure(figsize=(15, 15))
plt.suptitle('Univariate Analysis of Numerical Features', fontsize='20', fontweight='bold', alpha=0.8, y=1.)

for i, col in enumerate(numeric_features):
    plt.subplot(5, 3, i+1)
    sns.kdeplot(x=df_copy[col], fill=True, color='r')  # fill=True replaces shade=True
    plt.xlabel(col)
    plt.tight_layout()

plt.show()
```
